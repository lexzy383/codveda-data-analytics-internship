```python
# Work on a copy so the original stays untouched
clean = df.copy()

# Remove duplicate rows (none were found, kept as a safeguard)
clean = clean.drop_duplicates()

# Tidy column names: lowercase, underscores instead of spaces
clean.columns = clean.columns.str.strip().str.lower().str.replace(" ", "_")

# Convert Yes/No text to 1/0
clean["international_plan"] = clean["international_plan"].map({"Yes": 1, "No": 0})
clean["voice_mail_plan"] = clean["voice_mail_plan"].map({"Yes": 1, "No": 0})

# Convert True/False to 1/0
clean["churn"] = clean["churn"].astype(int)

# Make sure State is consistent text
clean["state"] = clean["state"].str.strip().str.upper()

clean.head()
```

```python
clean.info()
print("Rows before:", len(df), "| Rows after:", len(clean))
```

```python
clean.to_csv("cleaned_churn.csv", index=False)
```

Summary: The dataset has 667 rows and 20 columns. No missing values and no duplicate rows were found. I standardized column names, converted International plan, Voice mail plan and Churn into 1/0, and saved the result as cleaned_churn.csv
